```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/leontoddjohnson/datasets/refs/heads/main/data/coffee_analysis.csv"

df_coffee = pd.read_csv(url)

# View the first five rows
df_coffee.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import pickle

df_train, df_test = train_test_split(df_coffee, test_size=0.2, random_state=42)

X = df_train[["100g_USD"]]
y = df_train[["rating"]]

model_1 = LinearRegression()
model_1.fit(X, y)

with open('model_1.pickle', 'wb') as file:
    pickle.dump(model_1, file)
```

```python
from sklearn.tree import DecisionTreeRegressor

print(df_coffee['roast'].unique())

roast_cat = {
    'Light': 0,
    'Medium-Light': 1,
    'Medium': 2,
    'Medium-Dark': 3,
    'Dark': 4
}

df_coffee['roast_num'] = df_coffee['roast'].map(roast_cat)

X = df_train[['100g_USD', 'roast_num']]
y = df_train['rating']

model_2 = DecisionTreeRegressor()
model_2.fit(X, y)

with open('model_2.pickle', 'wb') as file:
    pickle.dump(model_2, file)
```
